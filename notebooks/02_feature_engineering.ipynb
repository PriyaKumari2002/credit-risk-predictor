```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("output.csv")
print(df.head(10))
```

```python
df.shape
```

```python
df['DAYS_EMPLOYED'].describe()
```

```python
df['DAYS_EMPLOYED'].replace(365243, np.nan, inplace=True)
df['YEARS_EMPLOYED'] = df['DAYS_EMPLOYED'].abs() / 365
print(df['YEARS_EMPLOYED'].describe())
```

```python
df['DEBT_INCOME_RATIO'] = df['AMT_CREDIT'] / df['AMT_INCOME_TOTAL']
df['DEBT_INCOME_RATIO'].describe()
```

```python
percentile_99 = df['DEBT_INCOME_RATIO'].quantile(0.99)
print(percentile_99)
```

```python
df['DEBT_INCOME_RATIO'] = df['DEBT_INCOME_RATIO'].clip(upper=percentile_99)
df['DEBT_INCOME_RATIO'].max()
```

```python
df['CREDIT_TERM'] = df['AMT_CREDIT'] / df['AMT_ANNUITY']
print(df['CREDIT_TERM'].describe())
```

```python
new_features = ['AGE_YEARS', 'YEARS_EMPLOYED', 'DEBT_INCOME_RATIO', 'CREDIT_TERM']
df[new_features].describe()
```

```python
df['CODE_GENDER'].value_counts()
```

```python
df['CODE_GENDER'] = df['CODE_GENDER'].map({'M': 0, 'F': 1, 'XNA': np.nan})
df['FLAG_OWN_CAR'] = df['FLAG_OWN_CAR'].map({'N': 0, 'Y': 1})
df['FLAG_OWN_REALTY'] = df['FLAG_OWN_REALTY'].map({'N': 0, 'Y': 1})
```

```python
df = df.dropna(subset=['CODE_GENDER'])
print(df.shape)
```

```python
cat_columns = ['NAME_CONTRACT_TYPE', 'NAME_TYPE_SUITE', 'NAME_INCOME_TYPE', 
               'NAME_EDUCATION_TYPE', 'NAME_FAMILY_STATUS', 'NAME_HOUSING_TYPE',
               'OCCUPATION_TYPE', 'WEEKDAY_APPR_PROCESS_START', 
               'ORGANIZATION_TYPE', 'EMERGENCYSTATE_MODE']

df = pd.get_dummies(df, columns=cat_columns)
print(df.shape)
```

```python
df.to_csv("Feature-Engg-output.csv", index=False)
print("CSV file saved!")
```
